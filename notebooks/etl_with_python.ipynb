```python
import pandas as pd

# Lê o CSV como DataFrame
df = pd.read_csv('customer_data.csv')

# Pega a coluna id como lista
user_ids = df['id'].tolist()
print(user_ids)

# Converte para lista de dicionários
users = df.to_dict(orient='records')

# Garante a estrutura esperada para a etapa de Transformação
for user in users:
    user['message'] = (
        f"Olá {user['name']}! "
        f"Seu cartão final {str(user['card_number'])[-4:]} "
        f"possui novidades exclusivas."
    )
```
